```python
corpus = """AI is amazing! It’s everywhere—smartphones, cars, even healthcare.
But wait, are we ready for its challenges? Think about privacy and fairness!"""

```

```python
print(corpus)
```

```python
## Tokenization
## convert a senetence into paragraph
from nltk.tokenize import sent_tokenize
sent_tokenize(corpus)
```

```python
documents = sent_tokenize(corpus)

for document in documents:
    print(document)
```

```python
## word toenizer
# paragraph -> words
# sentences -> words
from nltk.tokenize import word_tokenize
```

```python
## converting para or corpus into words
word_tokenize(corpus)
```

```python
## converting sentenc or documents into words
for document in documents:
    print(word_tokenize(document))
```

```python
from nltk.tokenize import wordpunct_tokenize, TreebankWordTokenizer
wordpunct_tokenize(corpus) ## it considers full stop(.) as seperate word
```

```python
TreebankWordTokenizer().tokenize(corpus) ## it doesn't consider full stop(.) as seperate word
```
